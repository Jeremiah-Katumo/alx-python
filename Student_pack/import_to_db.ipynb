```python
import sqlite3
import mysql.connector
from dotenv import load_dotenv
import os

load_dotenv()

sqlite_conn = sqlite3.connect("Maji_Ndogo_farm_survey_small.db")
sqlite_cur = sqlite_conn.cursor()

mysql_conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=os.getenv('PASSWORD'),
    database="maji_ndogo_farm_survey_db"
)
mysql_cur = mysql_conn.cursor()

# Example: migrate a table
sqlite_cur.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = sqlite_cur.fetchall()

for table in tables:
    print("Migrating:", table[0])
    # Fetch data
    sqlite_cur.execute(f"SELECT * FROM {table[0]}")
    rows = sqlite_cur.fetchall()

    placeholders = ", ".join(["%s"] * len(rows[0]))

    insert_query = f"INSERT INTO {table[0]} VALUES ({placeholders})"

    mysql_cur.executemany(insert_query, rows)
    mysql_conn.commit()
```

```python
import pandas as pd

dfs = pd.read_html("Maji_Ndogo_farm_survey_small.db")
df = dfs[0]   # first table

print(df.head())


import mysql.connector
import os
from dotenv import load_dotenv

load_dotenv()

mysql_conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=os.getenv('PASSWORD'),
    database="maji_ndogo_farm_survey_db"
)
cursor = mysql_conn.cursor()

# create insert query dynamically
cols = ", ".join(df.columns)
placeholders = ", ".join(["%s"] * len(df.columns))
sql = f"INSERT INTO survey_data ({cols}) VALUES ({placeholders})"

cursor.executemany(sql, df.values.tolist())
mysql_conn.commit()
```
